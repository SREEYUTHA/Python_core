```python
# CONCEPT - 1: Creating and accessing JSON data

import json


#==================================================================
data = {'name':'sreeyutha','age': '24','job':'AI Eng'}

text = json.dumps(data)

result = json.loads(text)
 
 
print(result)

print("The name of the person is:", result['name'])
print("The age of the person is:", result['age'])
print("The job of the person is:", result['job'])

#==================================================================
product = {
    "id": 101,
    "name": "Laptop",
    "price": 55000,
    "in_stock": True
}
# print("json.dumps:", text)
# print("json.loads:", result)
text2 = json.dumps(product)
result2 = json.loads(text2)

print(result2)

print("The ID of the product is:", result2['id']) 
print("The name of the product is:", result2['name'])
print("The price of the product is:", result2['price'])
print("Is the product in stock?", result2['in_stock'])
```

```python
#Concept-2: Serialization using json.dumps()

import json

data = {'name': 'sreeyutha', 'age': '24', 'job': 'AI Eng'}

text = json.dumps(data)

print("Serialized JSON data:", text)
print("Type of serialized data:", type(text))
```

```python
# Concept-3: Deserialization using json.loads()

data = {'name':'sreeyutha','age': '24','job':'AI Eng'}
text = json.dumps(data)
result = json.loads(text)
print(result)
print(type(result))
```

```python
# CONCEPT :4, Working with nested JSON

data = """
{
    "name": "sreeyutha",
    "age": 24,
    "job": "AI Eng",
    "address": {
        "street": "123 Main St",
        "city": "Hyderbad",
        "state": "CA"
    },
    "skills": ["Python", "Machine Learning", "Data Analysis"]
}
"""
result = json.loads(data)
print(result)
print(type(result))
print(result['address']['city'])
print(result['skills'][:])
```

```python
#CONCEPT-5: Working with JSON arrays

data = """
{
    "products":[
        {"id": 101, "name": "Laptop", "price": 55000},
        {"id": 102, "name": "Smartphone", "price": 25000},
        {"id": 103, "name": "Tablet", "price": 15000},
        {"id": 104, "name": "Smartwatch", "price": 10000}
    ]
}
"""

result = json.loads(data)
# print(result['products'][1]['name'])  # Accessing the name of the second product in the array
# print(type(result))

for i in result['products']:
    print(f"Products are: {i['name']} and their price is: {i['price']}")
```

```python
#Concept-6: Writing and reading files

info = {"name": "sreeyutha", "age": 24, "job": "AI Eng"}
with open('data1.json', 'w') as f:
    json.dump(info, f, indent = 6)


```

```python
#Concept-7: Reading JSON data from a file

with open('data1.json', 'r') as file:
    data = json.load(file)

print(data['job'])
```

```python
#Concept 7: Handling invalid JSON

data = '{"name": "sreeyutha", "age": 24, "job": "AI Eng"}'

try:
    result = json.loads(data)
except json.JSONDecodeError as e:
    print("Error decoding JSON:", e)
finally:
    print("Execution completed.")
```

# Example 2: Update an object inside a JSON array

```python

data = """
{
    "products":[
        {"id": 101, "name": "laptop", "price": 55000},
        {"id": 102, "name": "smartphone", "price": 25000},
        {"id": 103, "name": "tablet", "price": 15000},
        {"id": 104, "name": "smartwatch", "price": 10000}
    ]
}
"""

result = json.loads(data)
# print(result['products'][1]['name'])  # Accessing the name of the second product in the array
# print(type(result))
name = input("enter the item so price can increased by 50:").lower()

found = False

for i in result['products']:
    # print(f"Products are: {i['name']} and their price is: {i['price']}")
    if i['name'] == name:
        i['price'] += 50
        print(f"The price of {name} has been increased by 50. New price is: {i['price']}")
        found = True
        break

if not found:
    print(f"Item '{name}' not found.")

print(result)
    
```
